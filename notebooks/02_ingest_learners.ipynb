# Overview

1. Load `learners.csv` into source.learners once.
2. From then on, act as if the CSV never existed. As far as the pipeline knows, source.learners is the app's database table.

```sql
CREATE OR REPLACE TABLE language_app.source.raw_learners 
TBLPROPERTIES (
    'delta.enableChangeDataFeed' = 'true',
    'delta.enableRowTracking' = 'true',
    'delta.enableDeletionVectors' = 'true'
  )
AS
SELECT 
    learner_id
    , first_name
    , last_name
    , country
    , native_language
    , signup_date
    , subscription_tier
    , _rescued_data
FROM read_files(
  '/Volumes/language_app/source/raw_files/learners/learners.csv',
  format => 'csv',
  header => true
);
```
