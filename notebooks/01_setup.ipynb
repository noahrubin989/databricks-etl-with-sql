# Overview

The purpose of this notebook is to provide clear setup details that the rest of the project follows.

```python
%sql
CREATE CATALOG IF NOT EXISTS language_app;

-- Document the catalog so its purpose is discoverable from Catalog Explorer
-- (and via DESCRIBE CATALOG / information_schema calls).
COMMENT ON CATALOG language_app
  IS 'Catalog for the language learning app project: contains raw, processed, and reporting data used by the notebooks in this project.';

USE CATALOG language_app;

SELECT 'Catalog ''language_app'' exists, is documented, and is now the default catalog.' AS message;
```

```python
%sql
-- Create the schemas used across the language learning app project.
-- Bronze/Silver/Gold follows a medallion-style layout; 'source' holds external/raw inputs.

CREATE SCHEMA IF NOT EXISTS language_app.source;
COMMENT ON SCHEMA language_app.source
  IS 'External/raw source data ingested into the platform before any transformation.';

CREATE SCHEMA IF NOT EXISTS language_app.bronze;
COMMENT ON SCHEMA language_app.bronze
  IS 'Raw data as first ingested — append-only, unvalidated landing layer.';

CREATE SCHEMA IF NOT EXISTS language_app.silver;
COMMENT ON SCHEMA language_app.silver
  IS 'Cleansed, validated, and deduplicated data ready for analysis and feature work.';

CREATE SCHEMA IF NOT EXISTS language_app.gold;
COMMENT ON SCHEMA language_app.gold
  IS 'Business-level aggregates and reporting tables consumed by dashboards and apps.';

SELECT
  'Schemas source, bronze, silver, gold created and documented.' AS message;
```

```python
%sql
-- Create a Volume to hold raw input files from outside the platform.
-- Placed in the source schema, since these are raw external inputs.

CREATE VOLUME IF NOT EXISTS language_app.source.raw_files;

COMMENT ON VOLUME language_app.source.raw_files
  IS 'Raw input files (external source data) staged for the language learning app project.';

SELECT
  'Volume created at /Volumes/language_app/source/raw_files/' AS volume_path;
```
